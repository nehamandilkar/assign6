### bitwise Operator 


### and  

```python
print(3&6)
print(5&7)
print(8&4)
print(6&2)
print(9&3)
```

### or

```python
print(3|6)
print(5|7)
print(8|4)
print(6|2)
print(9|3)
```

### xor

```python
print(3^6)
print(5^7)
print(8^4)
print(6^2)
print(9^3)
```

### left shift 


```python
print(5<<2)
print(5<<7)
print(8<<4)
print(6<<2)
print(9<<3)
```

### right shift 

```python
print(5>>2)
print(5>>7)
print(8>>4)
print(6>>2)
print(9>>3)
```
